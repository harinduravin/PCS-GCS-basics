```python
import os
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm

output_folder = 'plot_outputs'

# Font setup
local_fonts_path = os.path.expanduser('~/.local/share/fonts')
fm.fontManager.addfont(os.path.join(local_fonts_path, 'times.ttf'))

plt.rcParams.update({
    "font.size": 12,
    "grid.linestyle": "--",
    "axes.grid": True
})

plt.rcParams['pdf.fonttype'] = 42
plt.rcParams['ps.fonttype'] = 42
plt.rcParams['font.family'] = 'Times New Roman'
plt.rcParams['mathtext.fontset'] = 'stix'

# -------------------------
# Helper to load a dataset
# -------------------------
def load_constellation(folder, snr=13, papr=None):
    if papr is None:
        fname = f"snr_{snr}_papr_No PAPR.npz"
    else:
        fname = f"snr_{snr}_papr_{papr}.npz"
    path = os.path.join(folder, fname)
    data = np.load(path)
    return data['constellation_points'], data['constellation_probabilities']

# -------------------------
# Folders and configurations
# -------------------------
base_120 = 'results_64_pn_120Hz_lowpn'
base_220 = 'results_64_pn_220Hz_lowpn'
base_120_g2 = 'results_64_pn_120Hz_lowpn_g2'
base_220_g2 = 'results_64_pn_220Hz_lowpn_g2'

# PAPR list for the first three columns; last column is "No PAPR"
papr_vals = [6.5, 7.0, 7.5]

# Load all constellations
# Row 0: 120 Hz, original
points_120 = []
probs_120 = []
for papr in papr_vals:
    p, pr = load_constellation(base_120, snr=13, papr=papr)
    points_120.append(p)
    probs_120.append(pr)
p_no, pr_no = load_constellation(base_120, snr=13, papr=None)
points_120.append(p_no)
probs_120.append(pr_no)

# Row 1: 220 Hz, original
points_220 = []
probs_220 = []
for papr in papr_vals:
    p, pr = load_constellation(base_220, snr=13, papr=papr)
    points_220.append(p)
    probs_220.append(pr)
p_no2, pr_no2 = load_constellation(base_220, snr=13, papr=8.0)
points_220.append(p_no2)
probs_220.append(pr_no2)

# Row 2: 120 Hz, g2
points_120_g2 = []
probs_120_g2 = []
for papr in papr_vals:
    p, pr = load_constellation(base_120_g2, snr=13, papr=papr)
    points_120_g2.append(p)
    probs_120_g2.append(pr)
p_no_g2, pr_no_g2 = load_constellation(base_120_g2, snr=13, papr=None)
points_120_g2.append(p_no_g2)
probs_120_g2.append(pr_no_g2)

# Row 3: 220 Hz, g2
points_220_g2 = []
probs_220_g2 = []
for papr in papr_vals:
    p, pr = load_constellation(base_220_g2, snr=13, papr=papr)
    points_220_g2.append(p)
    probs_220_g2.append(pr)
p_no2_g2, pr_no2_g2 = load_constellation(base_220_g2, snr=13, papr=None)
points_220_g2.append(p_no2_g2)
probs_220_g2.append(pr_no2_g2)

# -------------------------
# Plotting
# -------------------------
fig, axes = plt.subplots(4, 4, figsize=(12, 12))

# Collect all rows for easy looping
all_rows = [
    (points_120, probs_120),
    (points_220, probs_220),
    (points_120_g2, probs_120_g2),
    (points_220_g2, probs_220_g2),
]

for i, (points_row, probs_row) in enumerate(all_rows):
    for j in range(4):
        ax = axes[i, j]
        plot_constellation(ax, points_row[j], probs_row[j], None, 2.1)

# Column titles
col_titles = ["PAPR 6.5 dB", "PAPR 7.0 dB", "PAPR 7.5 dB", "W/O PAPR Limit"]
for j, title in enumerate(col_titles):
    axes[0, j].set_title(title, fontsize=15)

# Row titles
row_titles = [
    "120 Hz \n g = 1",
    "220 Hz \n g = 1",
    "120 Hz \n g = 2",
    "220 Hz \n g = 2",
]

for i, title in enumerate(row_titles):
    axes[i, 0].annotate(
        title,
        xy=(-0.35, 0.5),
        xycoords="axes fraction",
        rotation=90,
        va="center",
        ha="center",
        fontsize=15,
        fontweight="bold"
    )

plt.tight_layout()
# fig.savefig(os.path.join(output_folder, 'learned_constellations_grid.pdf'), format='pdf', dpi=fig.dpi, bbox_inches='tight')
plt.show()
```

```python
def plot_constellation_wlabels(points, probabilities, plot_title):

    points = np.asarray(points)
    probabilities = np.asarray(probabilities).flatten()

    # Scale marker sizes for visibility
    marker_sizes = 300 * probabilities / max(probabilities)

    plt.figure(figsize=(8, 8))
    plt.scatter(points.real, points.imag, s=marker_sizes, alpha=0.7, c='b', edgecolors='k')

    # Annotate with 6-bit binary labels
    for i, pt in enumerate(points):
        label = format(i, '06b')  # 6-bit binary string
        plt.text(pt.real + 0.05, pt.imag + 0.05, label, fontsize=9, ha='left', va='bottom', color='darkred')

    # Add grid and formatting
    plt.axhline(0, color='gray', linewidth=0.5, linestyle='--')
    plt.axvline(0, color='gray', linewidth=0.5, linestyle='--')
    plt.grid(True, linestyle=':')
    plt.xlabel("In-phase")
    plt.ylabel("Quadrature")
    plt.title(plot_title)
    plt.gca().set_aspect('equal', adjustable='box')
    plt.tight_layout()
    # plt.show()

def normalize_constellation(points, probabilities):

    # Compute the power of the constellation
    power = np.sum(probabilities * np.abs(points)**2)
    # Normalize the constellation points
    normalized_points = points / np.sqrt(power)
    return normalized_points
```

```python
norm_spiral64 = normalize_constellation(spiral64, np.ones([64])/64)

plot_constellation_wlabels(norm_spiral64, np.ones([64])/64,'Spiral')
```
